```python
df["Date"] = pd.to_datetime(df["Month"] + " " + df["Year"].astype(str))
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,5))

plt.plot(df["Date"], df["Rural"])

plt.xlabel("Date")
plt.ylabel("Rural CPI Index")
plt.title("Rural CPI Trend in India (2025-2026)")

plt.xticks(rotation=45)

plt.show()
```

```python
df["Inflation_Rate"] = df["Rural"].pct_change() * 100
```

```python
df.head()
```

```python
df["Inflation"] = df["Rural"].pct_change(periods=12) * 100

df.head(15)
```

```python
df[['Date','Rural','Inflation_Rate','Inflation']].tail()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.plot(df['Date'], df['Inflation'], marker='o')

plt.title("India Rural CPI Inflation Rate (2026)")
plt.xlabel("Date")
plt.ylabel("Inflation (%)")

plt.xticks(rotation=45)

plt.grid(True)

plt.show()
```

```python
# What this graph shows:
#X-axis → Time (months)
#Y-axis → Rural CPI Inflation (%)
#It shows how price levels changed over 2025–2026.
#The upward movement means rural prices increased over time.
#The sudden rise towards the end indicates acceleration in inflation.
#This is not just a graph; it tells an economic story.
```

```python
import plotly.graph_objects as go
```

```python
fig = go.Figure()

fig.add_trace(
    go.Indicator(
        mode="number",
        value=df["Inflation"].iloc[-1],
        title={"text": "Latest Inflation (%)"}
    )
)

fig.show()
```

```python
import plotly.graph_objects as go

fig = go.Figure()

fig.add_trace(go.Indicator(
    mode="number",
    value=round(df["Inflation"].iloc[-1],2),
    title={
        "text":"Latest Inflation",
        "font":{"size":14}
    },
    number={
        "suffix":"%",
        "font":{"size":35}
    },
    domain={'x':[0,0.25], 'y':[0,1]}
))

fig.add_trace(go.Indicator(
    mode="number",
    value=round(df["Inflation"].mean(),2),
    title={
        "text":"Average Inflation",
        "font":{"size":14}
    },
    number={
        "suffix":"%",
        "font":{"size":35}
    },
    domain={'x':[0.25,0.5], 'y':[0,1]}
))

fig.add_trace(go.Indicator(
    mode="number",
    value=round(df["Inflation"].max(),2),
    title={
        "text":"Highest Inflation",
        "font":{"size":14}
    },
    number={
        "suffix":"%",
        "font":{"size":35}
    },
    domain={'x':[0.5,0.75], 'y':[0,1]}
))

fig.add_trace(go.Indicator(
    mode="number",
    value=round(df["Inflation"].min(),2),
    title={
        "text":"Lowest Inflation",
        "font":{"size":14}
    },
    number={
        "suffix":"%",
        "font":{"size":35}
    },
    domain={'x':[0.75,1], 'y':[0,1]}
))


fig.update_layout(
    title="India Rural Inflation Dashboard",
    height=300,
    margin=dict(t=80,b=20,l=20,r=20)
)

fig.show()
```
